```python
from inspect import classify_class_attrs
filmes = ["Filme 1" , "Filme 2" , "Filme 3" , "Filme 4" , "Filme 5"]
print("Bem vindo a classificação do filme")
print("Temos cinco filmes para você classificar")
print("Digite 0 para sair")

for filme in filmes:
  classificacao = input(f"Como voce classifica '{filme}' de 1 a 5 (ou 0 para sair):")

  if classificacao == '0':
    print("Que pena")
    break

  classificacao = int(classificacao)
  if classificacao <1 or classificacao >5:
    print("Nota invalida")

  else:
    print(f"Você classificou '{filme}' com a '{classificacao}' estrela. \n")
print("Obrigado")




```
